## Create a minimum consistent segmentation

### Functions

```r
# Compute the per cell minimum segmentation
min_cell_segmentation_per_cell <- function(cnv_data, chroms, ids, bin_size, column_names){
  df <- data.frame(Col1 = character(), 
                   Col2 = character(), 
                   Col3 = integer(), 
                   Col4 = integer(), 
                   Col5 = integer(),
                   stringsAsFactors = FALSE) 
  for (id in ids) {
    for (chr in chroms) {
      temp <- cnv_data %>% dplyr::filter(chrom == chr & cell_id == id)
      
      start <- unique(temp$start)
      end <- start + bin_size - 1
      
      i <- 1
      st <- start[i]
      ed <- end[i]
      
      while (i < length(start)) { 
        t1 <- temp %>% filter(start == start[i])
        t2 <- temp %>% filter(start == start[i+1])
        
        if (t1$state == t2$state) {
          i <- i+1
          ed <- end[i]
          if (i == length(start)) {
            df <- rbind(df,c(id,chr,st,ed,t1$state))
          }
        } else {
          df <- rbind(df,c(id,chr,st,ed,t1$state))
          i <- i+1
          st <- start[i]
          ed <- end[i]  
          if (i == length(start)) {
            df <- rbind(df,c(id,chr,st,ed,t2$state))
          } 
        }        
      }
    }
  }
  
  
  colnames(df) <- column_names;
  return(df)
  
}


# Compute the minimum segmentation across all cells
min_consistent_segmentation <- function(cnv_data,bin_size,chroms,col_names) {
  df <- data.frame(Col1 = character(), 
                   Col2 = character(), 
                   Col3 = integer(), 
                   Col4 = integer(),
                   Col5 = integer(),
                   stringsAsFactors = FALSE) 
  ids <- unique(cnv_data$cell_id)
  
  for (chr in chroms) {
    print(chr)
    temp <- cnv_data %>% dplyr::filter(chrom == chr)
    start <- unique(temp$start)
    end <- start + bin_size - 1
    
    i <- 1
    st <- start[i]
    ed <- end[i]
    k <- 0
    
    while (i < length(start)) { 
      for (sample in ids) {
        
        t1 <- temp %>% filter(cell_id == sample & start == start[i])
        t2 <- temp %>% filter(cell_id == sample & start == start[i+1])
        
        if (t1$state == t2$state) {
          k <- k+1
          if (k == length(ids)){
            i <- i+1
            ed <- end[i]
            k <- 0
            
            if (i == length(start)){
              temp2 <- temp %>% filter(start == st & end == st + bin_size - 1)
              df <- rbind(df,cbind(ids,
                                   rep(sprintf("chr%s",chr), length(ids)),
                                   rep(st, length(ids)),
                                   rep(ed, length(ids)),
                                   temp2[order(match(temp2$cell_id,ids)),]$state))
              
            }
          }
        } else {
          temp2 <- temp %>% filter(start == st & end == st + bin_size - 1)
          df <- rbind(df,cbind(ids,
                               rep(sprintf("chr%s",chr), length(ids)),
                               rep(st, length(ids)),
                               rep(ed, length(ids)),
                               temp2[order(match(temp2$cell_id,ids)),]$state))
          
          k <- 0
          i <- i+1
          st <- start[i]
          ed <- end[i]
          
          if (i == length(start)){
            temp2 <- temp %>% filter(start == st & end == ed)
            df <- rbind(df,cbind(ids,
                                 rep(sprintf("chr%s",chr), length(ids)),
                                 rep(st, length(ids)),
                                 rep(ed, length(ids)),
                                 temp2[order(match(temp2$cell_id,ids)),]$state))
            
          }
          
          break
        }        
      }
    }   
  }
  
  colnames(df) <- col_names;
  return(df)
  
}


# Merge breakpoints of individual cells that are within distance max_gap from each other
# to retain segments of width >= mid_width
# segs: consistent segmentation across cells
# orig_segs: per cell segmentation 
merge_small_segments <- function(segs, orig_segs, min_width, max_gap) {
  last_length <- Inf
  
  #Special set of small segments that we want to be sure to keep
  small_keepers <- orig_segs[width(orig_segs) <= min_width]
  
  
  while (min(width(segs[[1]])) < min_width && last_length > length(segs[[1]])) {
    last_length <- length(segs[[1]])
    
    # Find the minimum width segment and merge it
    s <- segs[[1]]
    # keep segs that are below width threshold
    s <- s[width(s) < min_width]
    
    # remove segs from consideration overlapping small segments in the input
    # segmentation (these are often biologically relevant)
    overlaps_keepers <- subsetByOverlaps(s, small_keepers)
    s <- s[!s %in% overlaps_keepers]
    if (length(s) < 1) {
      break
    }
    min_idx_filtered <- which.min(width(s))
    min_idx <- which(segs[[1]] == s[min_idx_filtered])
    
    # Which segment do we want to merge the small segment into?
    target_idx <- NA
    
    # Decide which direction to merge into
    before_idx <- precede(segs[[1]][min_idx], segs[[1]])
    after_idx <- follow(segs[[1]][min_idx], segs[[1]])
    
    if (is.na(before_idx) & is.na(after_idx)) {
      break
    }
    
    if (is.na(before_idx) || distance(segs[[1]][min_idx], segs[[1]][before_idx]) > max_gap) {
      if (is.na(after_idx)) {
        break
      }
      target_idx <- after_idx
    } else if (is.na(after_idx) || distance(segs[[1]][min_idx], segs[[1]][after_idx]) > max_gap) {
      if (is.na(before_idx)) {
        break
      }
      target_idx <- before_idx
    } else {
      target_idx <- c(after_idx, before_idx)[which.min(width(segs[[1]][c(after_idx, before_idx)]))]
    }
    
    start_pos <- min(start(segs[[1]][c(min_idx, target_idx)]))
    end_pos <- max(end(segs[[1]][c(min_idx, target_idx)]))
    
    # Loop over all samples
    for (i in seq_len(length(segs))) {
      gr <- segs[[i]]
      
      # Adjust the target to include the small segment
      start(gr[target_idx]) <- start_pos
      end(gr[target_idx]) <- end_pos
      
      # Remove the small segment
      gr <- gr[-min_idx]
      
      # Update the list of segmentations
      segs[[i]] <- gr
    }
  }
  
  segs
  
}  

# Fill in gaps in the segmentation following merging segments
# segs: consistent copy number segmentation
# fill_gap_width: maximum gap width that is filled by merging into an adjacent segment 
merge_gaps <- function(segs, fill_gap_width) {
  last_length <- Inf
  merge_idx <- 0
  
  # Now fill in all of the gaps, if we want to do that
  gap_gr <- gaps(segs[[1]])
  keep <- !is.na(precede(gap_gr, segs[[1]])) & !is.na(follow(gap_gr, segs[[1]]))
  gap_gr <- gap_gr[keep]
  while (length(gap_gr) > 0 & (last_length > length(gap_gr))) {
    last_length <- length(gap_gr)
    merge_idx <- merge_idx + 1
    
    gap_idx <- which.min(width(gap_gr))
    gap <- gap_gr[gap_idx]
    if (width(gap) > fill_gap_width) {
      break
    }
    
    # Decide which direction to merge into
    before_idx <- precede(gap, segs[[1]])
    after_idx <- follow(gap, segs[[1]])
    
    if (is.na(before_idx) & is.na(after_idx)) {
      break
    }
    
    if (is.na(before_idx)) {
      target_idx <- after_idx
    } else if (is.na(after_idx)) {
      target_idx <- before_idx
    } else {
      target_idx <- c(after_idx, before_idx)[which.min(width(segs[[1]][c(after_idx, before_idx)]))]
    }
    
    start_pos <- min(c(start(gap), start(segs[[1]][target_idx])))
    end_pos <- max(c(end(gap), end(segs[[1]][target_idx])))
    
    # Loop over all samples
    for (i in seq_len(length(segs))) {
      # Adjust the target to include the gap region
      start(segs[[i]][target_idx]) <- start_pos
      end(segs[[i]][target_idx]) <- end_pos
    }
    
    gap_gr <- gaps(segs[[1]])
    keep <- !is.na(precede(gap_gr, segs[[1]])) & !is.na(follow(gap_gr, segs[[1]]))
    gap_gr <- gap_gr[keep]
  }
  
  segs
}


min_cell_segmentation_per_cell_segmented_data <- function(cnv_data, chroms, ids,column_names){
  df <- data.frame(Col1 = character(), 
                   Col2 = character(), 
                   Col3 = integer(), 
                   Col4 = integer(), 
                   Col5 = integer(),
                   stringsAsFactors = FALSE) 
  for (id in ids) {
    for (chr in chroms) {
      temp <- cnv_data %>% dplyr::filter(chrom == chr & cell_id == id)
      
      start <- unique(temp$start)
      end <- unique(temp$end)
      
      i <- 1
      st <- start[i]
      ed <- end[i]
      
      if (length(start) > 1){ 
        
        while (i < length(start)) { 
          t1 <- temp %>% filter(start == start[i])
          t2 <- temp %>% filter(start == start[i+1])
          
          if (t1$state == t2$state) {
            i <- i+1
            ed <- end[i]
            if (i == length(start)) {
              df <- rbind(df,c(id,chr,st,ed,t1$state))
            }
          } else {
            df <- rbind(df,c(id,chr,st,ed,t1$state))
            i <- i+1
            st <- start[i]
            ed <- end[i]  
            if (i == length(start)) {
              df <- rbind(df,c(id,chr,st,ed,t2$state))
            } 
          }        
        } 
          
      } else { 
          df <- rbind(df,c(id,chr,st,ed,temp$state))
          }
    }
  }
  
  
  colnames(df) <- column_names;
  return(df)
  
}

# Rebin the segmented CNA profiles
rebin <- function(bin_size,df){
  setDT(df)
  
  # Number of bins per segment
  df[, bins := (end - start + 1) / 500000]
  
  # Repeat rows according to bins
  df_expanded <- df[rep(seq_len(.N), bins)]
  
  # Create bin_start per expanded row
  df_expanded[, bin_start := start + 500000 * (seq_len(.N) - 1), by = .(cell_id, chrom, start)]
  
  # Create bin_end
  df_expanded[, bin_end := bin_start + 500000 - 1]
  
  # Drop helper column
  df_expanded[, bins := NULL]
  
}
```

## Identification of CNAs present in at least 1% of non-WGD cells and larger than 10Mb

```r
# Redo dice but filter for non-unique CNAs that are sufficiently large
df <- read_tsv("../Data/CNA_Data/Processed/CNA_Profiles_Segmented2.tsv")

df <- df %>%
  group_by(cell_id) %>%
  mutate(mean_state = mean(state, na.rm = TRUE)) %>%
  ungroup()

# define different cell groups (based on ploidy)
cell_ids_diploid <- df %>%
  filter(mean_state <= 3) %>%
  distinct(cell_id) %>%
  pull(cell_id)

cell_ids_wgd <- df %>%
  filter(mean_state <= 5 & mean_state >3) %>%
  distinct(cell_id) %>%
  pull(cell_id)

cell_ids_wgt <- df %>%
  filter(mean_state >5) %>%
  distinct(cell_id) %>%
  pull(cell_id)

# Get a minimum consistent segmentation per (diploid) cell - with segments, not bins
chroms <- c(sprintf("chr%s",1:22),"chrX","chrY")
col_names <- c("cell_id","chrom","start","end","state")
segs <- min_cell_segmentation_per_cell_segmented_data(
  df %>% filter(cell_id %in% cell_ids_diploid),
  chroms   = chroms,
  ids      = cell_ids_diploid,
  column_names = col_names
)

# filter segments that are present in at least 1% of cells and of length >= 10 Mb
segs <- segs %>%
  add_count(chrom, start, end, state) %>%   # adds column "n"
  mutate(start = as.numeric(start)) %>%
  mutate(end = as.numeric(end)) %>% 
  mutate(length = end-start) %>%
  filter((n >= floor(612*0.01) & length >= 10000000)) %>% # 612 = # "diploid" cells
  count(chrom, start, end, state, name = "count")

states <- sapply(chroms, function(chrom) {
  unique(segs[segs$chrom == chrom,]$state)
})

# more refinement of breakpoints for the sufficiently frequent + large segments
segs_mod <- c()
for (chr in chroms) {
  for (s in states[[chr]]){
    
    temp <- segs %>% filter(chrom == chr & state == s )
    
    dups_s <- temp$start[duplicated(temp$start)]
    dups_e <- temp$end[duplicated(temp$end)]
    
    segs_mod <- rbind(segs_mod, temp[! temp$start %in% dups_s & ! temp$end %in% dups_e,])
    
    if (length(dups_s) > 0){
      
      for (i in 1:length(dups_s)){
        ed_t <- temp[temp$start == dups_s[i],]$end
        d <- as.matrix(dist(ed_t))
        if (length(which((d > 2500000 | d == 0) == FALSE)) == 0){
          segs_mod <- rbind(segs_mod, temp[temp$start == dups_s[i],])
        } else if(length(which((d > 2500000 | d == 0) == FALSE)) > 0 & length(which((d > 2500000 | d == 0) == FALSE)) < length(d)){
          isolated <- ed_t[apply(d > 2500000 | d == 0, 1, all)]
          segs_mod <- rbind(segs_mod, temp[temp$start == dups_s[i] & ! temp$end %in% dups_e & temp$end %in% isolated,])
          
          non_isolated <- ed_t[! ed_t %in% isolated]
          max_c <- max(temp[temp$start == dups_s[i] & temp$end %in% non_isolated, ]$count)
          segs_mod <- rbind(segs_mod, temp[temp$start == dups_s[i] & temp$end %in% non_isolated & temp$count == max_c,])
          
        } else{
          max_c <- max(temp[temp$start == dups_s[i], ]$count)
          segs_mod <- rbind(segs_mod, temp[temp$start == dups_s[i] & temp$count == max_c,])
        }
      }
      
    } 
    
    if (length(dups_e) > 0){
      
      for (i in 1:length(dups_e)){
        st_t <- temp[temp$end == dups_e[i],]$start
        d <- as.matrix(dist(st_t))
        if (length(which((d > 2500000 | d == 0) == FALSE)) == 0){
          segs_mod <- rbind(segs_mod, temp[temp$end == dups_e[i] & ! temp$start %in% dups_s ,])
        } else if(length(which((d > 2500000 | d == 0) == FALSE)) > 0 & length(which((d > 2500000 | d == 0) == FALSE)) < length(d)){
          isolated <- st_t[apply(d > 2500000 | d == 0, 1, all)]
          segs_mod <- rbind(segs_mod, temp[temp$end == dups_e[i] & ! temp$start %in% dups_s & temp$start %in% isolated,])
          
          non_isolated <- st_t[! st_t %in% isolated]
          max_c <- max(temp[temp$end == dups_e[i] & temp$start %in% non_isolated, ]$count)
          segs_mod <- rbind(segs_mod, temp[temp$end == dups_e[i] & temp$start %in% non_isolated & temp$count == max_c,])
          
        } else {
          max_c <- max(temp[temp$end == dups_e[i], ]$count)
          segs_mod <- rbind(segs_mod, temp[temp$end == dups_e[i] & temp$count == max_c,])
        }
      }
      
    } 
    
    
  }
  
}


segs_mod <- data.frame(segs_mod)
segs_mod$start <- as.numeric(segs_mod$start)
segs_mod$end <- as.numeric(segs_mod$end)
segs_mod$chrom <- factor(segs_mod$chrom, chroms)
segs_mod <-segs_mod[order(segs_mod$chrom, segs_mod$start),] %>% distinct()

# Remove segments that are only >10Mb because they encompass a chromosome's centromere
centromeres <- data.frame(read.table("../Data/Other/Centromeres.txt",
                                     header = T))
segs_mod <- segs_mod %>% distinct(chrom,start,end,state,count) %>% mutate(len = end - start) 

len_new <- c()
for (i in 1:nrow(segs_mod)){
  cen_st <- centromeres[centromeres$chrom == segs_mod$chrom[i],]$chromStart
  cen_ed <- centromeres[centromeres$chrom == segs_mod$chrom[i],]$chromEnd
  
  st <- segs_mod$start[i] 
  ed <- segs_mod$end[i] 
  if (st <= cen_st & ed >= cen_ed ){
    len_new <- c(len_new, segs_mod$len[i] - (cen_ed-cen_st))
  } else if(st > cen_st & st < cen_ed & ed >= cen_ed){
    len_new <- c(len_new, segs_mod$len[i] - (cen_ed-st))
  } else if(st > cen_st & ed < cen_ed){
    len_new <- c(len_new, segs_mod$len[i] - (ed-st))
  } else if(st <= cen_st & ed > cen_st & ed < cen_ed){
    len_new <- c(len_new, segs_mod$len[i] - (ed-cen_st))
  } else if( cen_ed < st | cen_st > ed){
    len_new <- c(len_new, segs_mod$len[i])
  }
  
}
segs_mod <- segs_mod %>% mutate(len_mod = len_new) %>% filter((len_mod >= 10000000 & chrom != "chr7") |
                                                              (len_mod >= 12000000 & chrom != "chr7") | #there is a 11.5Mb gain around the centromere on chr7, which looked very unstable in single cell CNA calls - hence it was excluded.
                                                                      chrom == "chrY")


write_csv(segs_mod, "../Data/CNA_Data/Processed/CNASegments_10MbOrMore_1PerCentCellsOrMore.csv")
```

```r
# Import the final CNA segments
segs <- read_csv("../Data/CNA_Data/Processed/CNASegments_10MbOrMore_1PerCentCellsOrMore_Final.csv")

segs_distinct <- segs %>% distinct(chrom,start,end)

# Extract retained copy number bins from the original, segmented data
df <- read_tsv("../Data/CNA_Data/Processed/CNA_Profiles_Segmented_Original.tsv")

df_mod <- c()
setDT(df)
setDT(segs_distinct)

# Make sure columns are numeric
df[, `:=`(start = as.numeric(start), end = as.numeric(end))]
segs_distinct[, `:=`(start = as.numeric(start), end = as.numeric(end))]

# Give segment starts/ends distinct names for clarity
setnames(segs_distinct, c("start", "end"), c("seg_start", "seg_end"))

# Keys must match the join variables
setkey(df, chrom, start, end)
setkey(segs_distinct, chrom, seg_start, seg_end)

# Interval join: find bins fully contained in segments
tmp <- foverlaps(
  df, segs_distinct,
  by.x = c("chrom", "start", "end"),
  by.y = c("chrom", "seg_start", "seg_end"),
  type = "within",
  nomatch = 0L
)

# Fast mode function
Mode_fast <- function(x) {
  ux <- unique(x)
  ux[which.max(tabulate(match(x, ux)))]
}

# Collapse bins → modal state per segment × cell
df_mod <- tmp[, .(state = Mode_fast(state)), 
              by = .(cell_id, chrom, seg_start, seg_end)]

# Rename back to start/end if desired
setnames(df_mod, c("seg_start", "seg_end"), c("start", "end"))
setcolorder(df_mod, c("cell_id", "chrom", "start", "end", "state"))

# correct chrom order
df_mod[, chrom := factor(chrom, levels = c(sprintf("chr%s",1:22), "chrX","chrY"))]
setorder(df_mod, chrom, start, end)

segs_distinct[, chrom := factor(chrom, levels = c(sprintf("chr%s",1:22), "chrX","chrY"))]
setorder(segs_distinct, chrom, seg_start, seg_end)


# Clean the dataframe to only allow certain CNAs and get rid of noise

# Make sure diploid, WGD, WGT cell name vectors exist
setDT(df_mod); setDT(segs_distinct); setDT(segs)
if (!exists("cell_ids_diploid")) stop("Please define cell_ids_diploid (vector of diploid cells to update).")
if (!exists("cell_ids_wgd")) stop("Please define cell_ids_wgd (vector of WGD cells to update).")
if (!exists("cell_ids_wgt")) stop("Please define cell_ids_wgt (vector of WGT cells to update).")

# Normalize names in segs_new_diff -> 'start','end'
targets <- copy(segs_distinct)
if ("seg_start" %in% names(targets)) setnames(targets, "seg_start", "start")
if ("seg_end"   %in% names(targets)) setnames(targets, "seg_end",   "end")

# Coerce types to avoid join mismatches
df_mod[, chrom := as.character(chrom)]
targets[, chrom := as.character(chrom)]

df_mod[, `:=`(start = as.numeric(as.character(start)),
              end   = as.numeric(as.character(end)),
              state = as.numeric(as.character(state)))]
targets[, `:=`(start = as.numeric(as.character(start)),
               end   = as.numeric(as.character(end)))]

# Process the diploid cells

    # select cells 
    cells_to_process <- intersect(unique(cell_ids_diploid), unique(df_mod$cell_id))
    
    df_sub <- df_mod[cell_id %in% cells_to_process]
    cat("Rows in df_sub (cells to process):", nrow(df_sub), "\n")
    cat("Target segments (segs):", nrow(targets), "\n")
    
    # Keep only rows that match the target segments 
    df_targets <- merge(df_sub, targets, by = c("chrom","start","end"), all = FALSE)
    cat("Rows in df_mod that matched segs:", nrow(df_targets), "\n")
    if (nrow(df_targets) == 0L) stop("No rows matched segs for the requested cells. Check coordinates/chromosomes.")
    
    # Prepare segs_allowed: one row per allowed chrom:start-end:state (col name 'state_allowed')
    segs_allowed <- copy(segs)[, .(chrom, start, end, state_allowed = state)]
    
    # Attach allowed states (this will create multiple rows per df_target row when multiple allowed states exist)
    df_join <- merge(df_targets, segs_allowed, by = c("chrom","start","end"), allow.cartesian = TRUE)
    setDT(df_join)

    cat("Rows after attaching allowed states (duplicates possible):", nrow(df_join), "\n")
    
    # Compute new_state per (cell_id:chrom:start-end)
    df_join[, new_state := {
      # state : the current state from df_mod; state_allowed : candidate allowed states
      if (any(state == state_allowed)) {
        state                 # already allowed -> keep
      } else {
        diffs <- abs(state - state_allowed)
        cand  <- state_allowed[diffs == min(diffs)]
        if (length(cand) == 1L) cand else max(cand)   # tie-break -> max 
      }
    }, by = .(cell_id, chrom, start, end)]
    
    # Unique updates computed
    updates <- unique(df_join[, .(cell_id, chrom, start, end, new_state)])
    cat("Unique computed updates:", nrow(updates), "\n")
    
    # Bring in old_state from df_mod (single merge, avoids .x/.y trouble)
    old_states <- unique(df_mod[, .(cell_id, chrom, start, end, old_state = state)])
    cmp <- merge(old_states, updates, by = c("cell_id","chrom","start","end"), all = FALSE)
    
    # How many would actually change?
    cmp[, changed := (old_state != new_state)]
    n_diff <- sum(cmp$changed, na.rm = TRUE)
    cat("Number of rows where state WOULD change:", n_diff, "\n")
    
    if (n_diff == 0L) {
      cat("Nothing to apply: new_state equals old_state for all matched rows.\n")
    } else {
      updates_apply <- cmp[changed == TRUE, .(cell_id, chrom, start, end, new_state)]
      # apply updates in-place
      setkeyv(df_mod, c("cell_id","chrom","start","end"))
      setkeyv(updates_apply, c("cell_id","chrom","start","end"))
      df_mod[updates_apply, state := i.new_state]
      cat("Applied updates to df_mod. Rows changed:", nrow(updates_apply), "\n")
      # show examples
      print(head(updates_apply, 10))
    }

# WGD cells
    # limit to cells to process
    cells_to_process <- intersect(unique(cell_ids_wgd), unique(df_mod$cell_id))
    
    df_sub <- df_mod[cell_id %in% cells_to_process]
    cat("Rows in df_sub (cells to process):", nrow(df_sub), "\n")
    cat("Target segments (segs_new_diff):", nrow(targets), "\n")
    
    # Keep only rows that match the target segments (same behaviour as original loop)
    df_targets <- merge(df_sub, targets, by = c("chrom","start","end"), all = FALSE)
    cat("Rows in df_mod that matched segs_new_diff:", nrow(df_targets), "\n")
    if (nrow(df_targets) == 0L) stop("No rows matched segs_new_diff for the requested cells. Check coordinates/chromosomes.")
    
    
    # Prepare segs_allowed: one row per allowed state (col name 'state_allowed')
    segs_allowed <- copy(segs)[, .(chrom, start, end, state_allowed = state*2)]
    
    # Attach allowed states (this will create multiple rows per df_target row when multiple allowed states exist)
    df_join <- merge(df_targets, segs_allowed, by = c("chrom","start","end"), allow.cartesian = TRUE)
    
    cat("Rows after attaching allowed states (duplicates possible):", nrow(df_join), "\n")
    
    # Compute new_state per (cell_id,chrom,start,end)
    df_join[, new_state := {
      # state : the current state from df_mod; state_allowed : candidate allowed states
      if (any(state == state_allowed)) {
        state                 # already allowed -> keep
      } else {
        diffs <- abs(state - state_allowed)
        cand  <- state_allowed[diffs == min(diffs)]
        if (length(cand) == 1L) cand else max(cand)   # tie-break -> max
      }
    }, by = .(cell_id, chrom, start, end)]
    
    # Unique updates computed
    updates <- unique(df_join[, .(cell_id, chrom, start, end, new_state)])
    cat("Unique computed updates:", nrow(updates), "\n")
    
    # Bring in old_state from df_mod (single merge, avoids .x/.y trouble)
    old_states <- unique(df_mod[, .(cell_id, chrom, start, end, old_state = state)])
    cmp <- merge(old_states, updates, by = c("cell_id","chrom","start","end"), all = FALSE)
    
    # How many would actually change?
    cmp[, changed := (old_state != new_state)]
    n_diff <- sum(cmp$changed, na.rm = TRUE)
    cat("Number of rows where state WOULD change:", n_diff, "\n")
    
    if (n_diff == 0L) {
      cat("Nothing to apply: new_state equals old_state for all matched rows.\n")
    } else {
      updates_apply <- cmp[changed == TRUE, .(cell_id, chrom, start, end, new_state)]
      # apply updates in-place
      setkeyv(df_mod, c("cell_id","chrom","start","end"))
      setkeyv(updates_apply, c("cell_id","chrom","start","end"))
      df_mod[updates_apply, state := i.new_state]
      cat("Applied updates to df_mod. Rows changed:", nrow(updates_apply), "\n")
      # show examples
      print(head(updates_apply, 10))
    }

# WGT cells 
    # limit to cells to process
    cells_to_process <- intersect(unique(cell_ids_wgt), unique(df_mod$cell_id))
    
    df_sub <- df_mod[cell_id %in% cells_to_process]
    cat("Rows in df_sub (cells to process):", nrow(df_sub), "\n")
    cat("Target segments (segs_new_diff):", nrow(targets), "\n")
    
    # Keep only rows that match the target segments (same behaviour as original loop)
    df_targets <- merge(df_sub, targets, by = c("chrom","start","end"), all = FALSE)
    cat("Rows in df_mod that matched segs_new_diff:", nrow(df_targets), "\n")
    if (nrow(df_targets) == 0L) stop("No rows matched segs_new_diff for the requested cells. Check coordinates/chromosomes.")
    
    
    # Prepare segs_allowed: one row per allowed state (col name 'state_allowed')
    segs_allowed <- copy(segs)[, .(chrom, start, end, state_allowed = state*3)]
    
    # Attach allowed states (this will create multiple rows per df_target row when multiple allowed states exist)
    df_join <- merge(df_targets, segs_allowed, by = c("chrom","start","end"), allow.cartesian = TRUE)
    
    cat("Rows after attaching allowed states (duplicates possible):", nrow(df_join), "\n")
    
    # Compute new_state per (cell_id,chrom,start,end)
    df_join[, new_state := {
      # state : the current state from df_mod; state_allowed : candidate allowed states
      if (any(state == state_allowed)) {
        state                 # already allowed -> keep
      } else {
        diffs <- abs(state - state_allowed)
        cand  <- state_allowed[diffs == min(diffs)]
        if (length(cand) == 1L) cand else max(cand)   # tie-break -> max
      }
    }, by = .(cell_id, chrom, start, end)]
    
    # Unique updates computed
    updates <- unique(df_join[, .(cell_id, chrom, start, end, new_state)])
    cat("Unique computed updates:", nrow(updates), "\n")
    
    # Bring in old_state from df_mod (single merge, avoids .x/.y trouble)
    old_states <- unique(df_mod[, .(cell_id, chrom, start, end, old_state = state)])
    cmp <- merge(old_states, updates, by = c("cell_id","chrom","start","end"), all = FALSE)
    
    # How many would actually change?
    cmp[, changed := (old_state != new_state)]
    n_diff <- sum(cmp$changed, na.rm = TRUE)
    cat("Number of rows where state WOULD change:", n_diff, "\n")
    
    if (n_diff == 0L) {
      cat("Nothing to apply: new_state equals old_state for all matched rows.\n")
    } else {
      updates_apply <- cmp[changed == TRUE, .(cell_id, chrom, start, end, new_state)]
      # apply updates in-place
      setkeyv(df_mod, c("cell_id","chrom","start","end"))
      setkeyv(updates_apply, c("cell_id","chrom","start","end"))
      df_mod[updates_apply, state := i.new_state]
      cat("Applied updates to df_mod. Rows changed:", nrow(updates_apply), "\n")
      # show examples
      print(head(updates_apply, 10))
    }



chroms <- c(sprintf("chr%s",c(1:22)),"chrX","chrY")
df_mod[, chrom := factor(chrom, levels = chroms)]
setorder(df_mod, chrom, start, end)

write_tsv(df_mod, "../Data/CNA_Data/Processed/CNA_Profiles_Segmented_10Mb_1PerCent.tsv")
```

## Single-cell phylogenetic reconstruction using maximum parsimony

```r
# Load the CNA data 
df <- read_tsv("../Data/CNA_Data/Processed/CNA_Profiles_Segmented_10Mb_1PerCent.tsv")

# Format the copy number data
cna_data <- rebin(500000,df) %>% dplyr::mutate(Location = sprintf("%s:%s-%s",chrom,bin_start,bin_end))

# Format cna data for heatmap plot
cna_df <- cna_data[,c("cell_id","state","Location")] %>% tidyr::pivot_wider(names_from = Location,
                                                                            values_from = `state`)
rownames(cna_df) <- cna_df$cell_id

cna_mat <- as.matrix(cna_df[,-1])
rownames(cna_mat) <- rownames(cna_df)

# Create a dummy diploid
diploid <- cna_mat[1,]
diploid[-grep("chrX|chrY",names(diploid))] <- rep(2,length(diploid[-grep("chrX|chrY",names(diploid))]))
diploid[grep("chrX|chrY",names(diploid))] <- rep(1,length(grep("chrX|chrY",names(diploid))))

# Add the dummy diplot
cna_mat_p <- rbind(diploid,cna_mat)

# Create phylogenetic data object for practchet
pratchet_data <- phyDat(cna_mat_p, type = "USER", levels = c(0:12))

# Set random seed for reproducibility
set.seed(7)

# Obtain maximum parsimony tree
treeRatchet  <- pratchet(data = pratchet_data, 
                         trace = 0, minit=100,maxit = 500000000, k = 200)
# Assign edge lengths
treeRatchet  <- acctran(treeRatchet, pratchet_data)

# Remove edges of length 0
treeRatchet  <- di2multi(treeRatchet)

tree <- ape::root(treeRatchet,outgroup = "diploid", edgelabel = T) 
write.tree(tree,"../Data/CNA_Data/PhylogeneticTrees/SingleCell_MaximumParsimony_Tree.nwk")
```

```r
# Load the CNA data 
df <- read_tsv("../Data/CNA_Data/Processed/CNA_Profiles_Segmented_10Mb_1PerCent.tsv")

# Format the copy number data
cna_data <- rebin(500000,df) %>% dplyr::mutate(Location = sprintf("%s:%s-%s",chrom,bin_start,bin_end))

# Format cna data for heatmap plot
cna_df <- cna_data[,c("cell_id","state","Location")] %>% tidyr::pivot_wider(names_from = Location,
                                                                            values_from = `state`)
rownames(cna_df) <- cna_df$cell_id

cna_mat <- as.matrix(cna_df[,-1])
rownames(cna_mat) <- rownames(cna_df)

#Order cells according to tree
# Retrieve the order of cells in the phylogenetic tree plotted above 'mcc'
order <- get_taxa_name(mcc)

# Remove the dummy diploid cell
order <- order[order != "diploid"]

# Order cells according to the phylogenetic tree plotted above 'mcc'
cna_mat <- cna_mat[order,]

# k means clustering
km.out <- kmeans(cna_mat, centers = 8, nstart = 20)
clusters <- data.frame("Cell_id" = names(km.out$cluster)) %>%
  mutate(Clade = c(rep(1,8), 
                         rep(2,48),
                         rep(3,24),
                         rep(4,6),
                         rep(5,9),
                         rep(6,8),
                         rep(7,27),#
                         rep(8,155),
                         rep(9,8),#
                         rep(10,72),#
                         rep(11,25),#
                         rep(12,13), #
                         rep(13,15),#
                         rep(14,13),#
                         rep(15,5),#
                         rep(16,7),#
                         rep(17,48),#
                         rep(18,55),#
                         rep(19,25),#
                         rep(20,35),#
                         rep(21,14),#
                         rep(22,6),#
                         rep(23,7),#
                         rep(24,14),#
                         rep(25,15),
                         rep(26,6))) %>% 
  mutate(SuperClade = case_when(Clade %in% c(1,2) ~ 1,
                               Clade %in% c(4,5) ~ 2,
                               Clade %in% c(12,13) ~ 3,
                               Clade %in% c(14,17,19,20) ~ 4,
                               Clade %in% c(16) ~ 5,
                               Clade %in% c(18) ~ 6,
                               Clade %in% c(3,15,8) ~ 7,
                               Clade %in% c(9) ~ 8,
                               Clade %in% c(10,11,6) ~ 9,
                               Clade %in% c(7) ~ 10,
                               Clade %in% c(25) ~ 11,
                               Clade %in% c(21,22,23,24,26) ~ 12))



write_csv(clusters,"../Data/CNA_Data/Processed/Genetic_Clades_And_SuperClades.csv")

# Row annotation
row_annotation_l <- data.frame(clusters$SuperClade,clusters$Clade)
colnames(row_annotation_l) <- c("Super clade","Genetic clade")
rownames(row_annotation_l) <- rownames(cna_mat)

row_annotation_l$`Genetic clade` <- factor(row_annotation_l$`Genetic clade`, 
                                           levels = sprintf("%s",1:26))
row_annotation_l$`Super clade` <- factor(row_annotation_l$`Super clade`, 
                                           levels = sprintf("%s",1:12))

pal32 <- c(colour("bright")(7), colour("vibrant")(7),colour("light")(9),colour("muted")(9))
my_cols <- pal32[-c(3,4,5,7,17,22,30)]
my_cols <- c(my_cols[c(1:5,7,6,8:12,25,14:24,13)],"#eeba0b")

row_annotation_l_h <- ComplexHeatmap::rowAnnotation(df = row_annotation_l,
                                                    col = list(`Genetic clade` = setNames(my_cols, unique(clusters$Clade)),
                                                               `Super clade` = c("1" = "#4e0921",
                                                                                "2" = "#9d0208",
                                                                                "3" = "#d00000",
                                                                                "4" = "#e85d04",
                                                                                "5" = "#faa307",
                                                                                "6" = "#ffc83b",
                                                                                "7" = "#e3f2fd",
                                                                                "8" = "#c8d7f9",
                                                                                "9" = "#7ea3f1",
                                                                                "10" = "#487bea",
                                                                                "11" = "#1957db",
                                                                                "12" = "#023087")),
                                                    simple_anno_size = unit(1, "cm"),
                                                    annotation_name_side = "top"
)

sample <- unlist(str_split(rownames(cna_mat),"_DATAC"))[seq(1,2*nrow(cna_mat),2)]
sample[grep("POTm",sample)] <- "T0.A"

sample[grep("POTp",sample)] <- "T0.B"
sample[grep("DSIa",sample)] <- "T1.A"
sample[grep("DSIb",sample)] <- "T1.B"
sample[grep("DSIVa",sample)] <- "T4.A"
sample[grep("DSIVb",sample)] <- "T4.B"
sample[grep("DT1_P1a",sample)] <- "T11.A"
sample[grep("DT1_P1b",sample)] <- "T12.B"

sample <- factor(sample,levels = c("T0.A","T0.B","T1.A","T1.B","T4.A","T4.B","T11.A","T12.B"))

time <- unlist(str_split(rownames(cna_mat),"_DATAC"))[seq(1,2*nrow(cna_mat),2)]
time[grep("POT",time)] <- "T0"
time[grep("DSIa|DSIb",time)] <- "T1"
time[grep("DSIV",time)] <- "T4"
time[grep("DT1",time)] <- "T11.5"

time <- factor(time,levels = c("T0","T1","T4","T11.5"))

clusters$Sample <- sample
clusters$Time <- time

epi_class <- data.frame(read_csv("../Data/ATAC_Data/GeneExpressionScores/DATAC/DATAC_Consensus_EpigeneticClasses_C4.csv"))

rownames(epi_class) <- epi_class$Cell_id
epi_class <- epi_class[clusters$Cell_id,]

row_annotation_r <- data.frame(sample, time, epi_class$consensus_cluster)
colnames(row_annotation_r) <- c("Sample", "Time", "Epigenetic class")
rownames(row_annotation_r) <- rownames(cna_mat)

#row_annotation_r <- row_annotation_r[rownames(row_annotation_r) %in% rownames(row_annotation_l),]
row_annotation_r_h <- ComplexHeatmap::rowAnnotation(df = row_annotation_r,
                                                    
                                                    col = list( `Sample` = c("T0.A" = "#730d5d",
                                                                            "T0.B" = "#b81595",
                                                                            "T1.A" =  "#f077d6",
                                                                            "T1.B" =  "#f8bceb",
                                                                            "T4.A" = "#c9bcf8",
                                                                            "T4.B" = "#7e60ef",
                                                                            "T11.A"= "#a25fef",
                                                                            "T12.B"= "#6113bc"),
                                                                `Time` =  c("T0" = "#961179",
                                                                                      "T1" =  "#F49AE1",
                                                                                      "T4" = "#A48EF4",
                                                                                      "T11.5"= "#8239D6"),
                                                                `Epigenetic class` =  c("\u03B1" = "#ebf2cf",
                                                                                        "\u03B2" = "#a7c957",
                                                                                        "\u03B3" = "#6a994e",
                                                                                        "\u03B4" = "#2f5637")),
                                                    annotation_legend_param = list(
                                                      `Epigenetic class` = list(
                                                        at = c("\u03B1", "\u03B2", "\u03B3", "\u03B4"),
                                                        labels = c(expression(alpha), expression(beta), expression(gamma), expression(delta))
                                                      )
                                                    ),
                                                    simple_anno_size = unit(1, "cm"),
                                                    
                                                    annotation_name_side = "top"
)


# Column annotation to separate chromosomes, listed in order 1,2,3,...,22,X
column_annotation <- data.frame(unlist(strsplit(colnames(cna_mat), split=':', fixed=TRUE))[seq(1,length(colnames(cna_mat))*2,by=2)])
colnames(column_annotation) <- "Chromosome"
rownames(column_annotation) <- colnames(cna_mat)
column_annotation$Chromosome <- factor(column_annotation$Chromosome, levels = c(sprintf("chr%s",c(1:22)),"chrX","chrY"))

column_annotation_h <- ComplexHeatmap::HeatmapAnnotation(df = column_annotation,
                                                         col = list(Chromosome = c("chr1"="white", "chr2"="black", "chr3"="white", "chr4"="black",
                                                                                   "chr5"="white", "chr6"="black","chr7"="white", "chr8"="black",
                                                                                   "chr9"="white", "chr10"="black","chr11"="white", "chr12"="black",
                                                                                   "chr13"="white", "chr14"="black","chr15"="white", "chr16"="black",
                                                                                   "chr17"="white", "chr18"="black","chr19"="white", "chr20"="black",
                                                                                   "chr21"="white", "chr22"="black","chrX"="white","chrY"="black") ),
                                                         simple_anno_size = unit(0.5, "cm")
)

# Define heatmap colours

myColor <- c("#3399db","#85c1e9","#d3d3d3","#fdc38d","#fda16c","#f57d52",
             "#e7523a","#cf2518","#ad0000","#c4188a","#a7007c","#87027a")
myBreaks <- seq(0,11,1)

colours <- circlize::colorRamp2(myBreaks,myColor) 

# Plot heatmap
h <- ComplexHeatmap::Heatmap(as.matrix(cna_mat), 
                             col = colours,
                             bottom_annotation = column_annotation_h,
                             left_annotation = row_annotation_l_h,
                             right_annotation = row_annotation_r_h,
                             cluster_columns = F,
                             cluster_rows = F,
                             show_column_names = FALSE,
                             show_row_names =F,
                             row_split = clusters$Clade_final,
                             heatmap_legend_param = list(title = "Copy number",
                                                         at = myBreaks),
                             raster_quality = 10,
                             width = unit(20,"cm"),
                             height = unit(30,"cm"))


# Draw the heatmap
options(repr.plot.width=12, repr.plot.height=14)
draw(h, heatmap_legend_side = "right", annotation_legend_side = "right")
```

## Clade phylogeny using modal copy number profiles

```r
# Load the CNA data 
df <- read_tsv("../Data/CNA_Data/Processed/CNA_Profiles_Segmented_10Mb_1PerCent.tsv")

# Format the copy number data
cna_data <- rebin(500000,df) %>% dplyr::mutate(Location = sprintf("%s:%s-%s",chrom,bin_start,bin_end))

# Format cna data for heatmap plot
cna_df <- cna_data[,c("cell_id","state","Location")] %>% tidyr::pivot_wider(names_from = Location,
                                                                            values_from = `state`)
rownames(cna_df) <- cna_df$cell_id

# Import clades
clades <- read_csv("../Data/CNA_Data/Processed/Genetic_Clades_And_SuperClades.csv")
rownames(clades) <- clades$Cell_id

mode_df <- c()
for (c in sprintf("%s",1:length(unique(clades$Clade)))){
  cells <- clades[clades$Clade == c,]$Cell_id
  temp <- data.frame(cna_df[cna_df$cell_id %in% cells,2:ncol(cna_df)])
  mode_cn <- c()
  for (i in 1:ncol(temp)) {
    mode_cn[i] <- pracma::Mode(temp[,i])
  }
  mode_df <- rbind(mode_df, cbind(c,t(as.numeric(mode_cn))))
}

rownames(mode_df) <- mode_df[,1]
colnames(mode_df) <- colnames(cna_df)

mode_df <- mode_df[,-1]
mode_df_num <- apply(mode_df,2, as.numeric)
rownames(mode_df_num ) <- rownames(mode_df)

# Add a dummy diploid 
diploid <- mode_df_num[1,]
diploid[-grep("chrX|chrY",names(diploid))] <- rep(2,length(diploid[-grep("chrX|chrY",names(diploid))]))
diploid[grep("chrX|chrY",names(diploid))] <- rep(1,length(grep("chrX|chrY",names(diploid))))

mode_df_num_p <- rbind(diploid,mode_df_num)

# Create phylogenetic data object for practchet
pratchet_data <- phyDat(mode_df_num_p, type = "USER", levels = c(0:12))

# Set random seed for reproducibility
set.seed(7)

# Obtain maximum parsimony tree
treeRatchet  <- pratchet(data = pratchet_data, 
                         trace = 0, minit=100,maxit = 500000000, k = 200)

# Assign edge lengths
treeRatchet  <- acctran(treeRatchet, pratchet_data)

# Remove edges of length 0
treeRatchet  <- di2multi(treeRatchet)

# Obtain final tree
tree <- ape::root(treeRatchet,outgroup = "diploid", edgelabel = T) 

# Save tree
write.tree(tree,"../Data/CNA_Data/PhylogeneticTrees/Clades_MaximumParsimony_Tree.nwk")
```
