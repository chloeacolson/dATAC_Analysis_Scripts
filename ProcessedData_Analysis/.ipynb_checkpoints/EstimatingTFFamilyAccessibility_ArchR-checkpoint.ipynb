# Predicting transcription factor family scores from the ATAC data

```r
library(ArchR)
library(BSgenome.Hsapiens.UCSC.hg38)
library(readr)
library(reticulate)
```

```r
# Load ArchR project

proj <- loadArchRProject(path = "Data/ArchR_Projects/Save_SW620_DATAC_Results",
                         force = FALSE)
```

```r
# Add TF motif annotations to the ArchR project
proj <- addMotifAnnotations(ArchRProj = proj, 
                                motifSet = "cisbp",
                                annoName = "Motif_cisbp",
                                force = T)

# Add a set of background peaks to be used for bias-correction of the per cell TF deviation scores
proj <- addBgdPeaks(proj)
```

```r
# Add a deviation matrix to the ArchR project
proj <- addDeviationsMatrix(
  ArchRProj = proj, 
  peakAnnotation = "Motif_cisbp",
  force = TRUE
)
```

```r
# Get the TF enrichment matrices from Arch R
motifs <- getMatrixFromProject(proj, useMatrix = "Motif_cisbpMatrix")

# Extract the z-score matrix
motif_mat_z <-  as.matrix(motifs@assays@data@listData$z)

# Edit the cell ids to match the CNA data
dsiv_matchings <- read_csv("Data/Cell_Lists/SW620bc_DSIV_barcode_matching.csv",
                           col_types = "cc")

cnv_ids <- colnames(motif_mat_z)
for (id in colnames(motif_mat_z)){
  if (id %in% dsiv_matchings$cell_barcode){
    cnv_ids[cnv_ids == id] = dsiv_matchings[dsiv_matchings$cell_barcode == id,]$sample_name
  } 
}

cnv_ids <- str_replace(cnv_ids,"SW620bc_DSIb_DATAC#","")
cnv_ids <- str_replace(cnv_ids,"SW620bc_POTa_DATAC#","")
cnv_ids <- str_replace(cnv_ids,"SW620bc_DT1_P1a_DATAC#","")
cnv_ids <- str_replace(cnv_ids,"SW620bc_DT1_P1b_DATAC#","")
cnv_ids <- str_replace(cnv_ids,"_ATAATGTG","")
cnv_ids <- str_replace(cnv_ids,"_TCGATATC","")
cnv_ids <- str_replace(cnv_ids,"_TACTCATA","")

colnames(motif_mat_z) <- cnv_ids

# Save the project
#saveArchRProject(ArchRProj = proj, outputDirectory = "Data/ArchR_Projects/Save_SW620_DATAC_Results", load = FALSE)
```

```r
# Format the TF z-score matrix as input for MAGIC:
# Columns = TFs
# Rows = cells

# Take the transpose of the filtered TF score matrix and turn into dataframe
motif_mat_z_df <- as.data.frame(t(as.matrix(motif_mat_z)))

# Name columns using TF names
colnames(motif_mat_z_df) <- rownames(motif_mat_z)

# Save the TF z-score matrix
#write_csv(cbind("Cell_id" = rownames(motif_mat_z_df),motif_mat_z_df), "Data/ATAC_Data/TFScores/TF_Zscores_Cisbp.csv")


# Python package needed to run magic through R
# R environment file specifying not to autoconfigure reticulate and automatically create a virtual environment
readRenviron("~/.Renviron")

py_config()

# Make sure R finds magic
py_discover_config(required_module = "magic")

# Import magic in python
import("magic")

# Import magic R library
library(Rmagic)

# Run magic on gene_df - note that magic has randomness to it and, therefore, generates slightly different outputs everytime, i.e.,
# results that follow depend on our gene score matrix post-magic
magic_output <- magic(motif_mat_z_df, t = 2, seed = 7)
motif_mat_z_df_m <- magic_output$result

write_csv(cbind("Cell_id" = rownames(motif_mat_z_df_m),motif_mat_z_df_m), "Data/ATAC_Data/TFScores/TF_Zscores_Cisbp_Magic_t2.csv")
```

```r
families <- read_csv("Data/ATAC_Data/TFScores/TF_Families.csv", col_types = "cc")

TFs <- colnames(motif_mat_z_df_m)
TFs <- unlist(str_split(TFs,"_",2))[seq(1,length(unlist(str_split(TFs,"_",2))),2)]
colnames(motif_mat_z_df_m) <- TFs

tfbs_fams <- c()
fams <- c()
for (family in unique(families$Family)){
  TFs_temp <- TFs[which(TFs %in% families[families$Family == family,]$TF)]
  num <- length(TFs_temp)
  if (num > 0){
    if (num > 1){
      zscores <- as.numeric(rowSums(motif_mat_z_df_m[,TFs_temp])/sqrt(num))
      fams <- c(fams, family)
      tfbs_fams <- rbind(tfbs_fams,zscores)
    } else if  (num == 1){
      fams <- c(fams, family)
      tfbs_fams <- rbind(tfbs_fams,motif_mat_z_df_m[,TFs_temp])
    }
  }
  
}
rownames(tfbs_fams) <- fams
colnames(tfbs_fams) <- rownames(motif_mat_z_df_m)

tfbs_fams <- cbind("Family"= rownames(tfbs_fams),tfbs_fams)
write_csv(data.frame(tfbs_fams),"Data/ATAC_Data/TFScores/TF_Family_ZScores_Cisbp_Magic_t2.csv")
```
